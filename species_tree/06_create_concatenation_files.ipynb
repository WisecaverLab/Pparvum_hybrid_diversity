# Prepare files for constructing a species tree


```python
from Bio import SeqIO
import subprocess
```

```python
rootdir="hybrid_orthofinder/"
ogalndir=rootdir+"19_trimal/"
inparamfile=rootdir+"22_concatenated/Supermatrix_in.param"
inalnfile=rootdir+"22_concatenated/Supermatrix_in.aln"
```

```python
lscmd = 'ls ' + ogalndir + '*.trim.aln'
files=subprocess.check_output(lscmd, shell=True).decode().split("\n")
files.pop()
```

```python
scogs=[]
for file in files:
    file=file.split("/")[-1]
    file=file.split(".")[0]
    scogs.append(file)
```

```python
print(len(scogs))
```

```python
seqDict = {}
counter = 0
for og in scogs:
    ogalnfile = ogalndir + og + '.trim.aln'
    appendtoggle=True
    for seq_record in SeqIO.parse(ogalnfile, "fasta"):
        sequence = str(seq_record.seq)
        strain = seq_record.id.split('_')[-1]

        if len(sequence) > 150:
        #print(strain)
            if appendtoggle==True:
                if strain not in seqDict:
                    seqDict[strain] = {}

                seqDict[strain][og] = sequence
        else:
            appendtoggle=False
print(len(seqDict.keys()))
```

```python
fo = open(inparamfile, 'w')
start = 0
first = next(iter(seqDict.keys()))
for og in seqDict[first]:
    start += 1
    end = int(start + len(seqDict[first][og]) - 1)
    fo.write('DNA, ' + og + ' = ' + str(start) + '-' + str(end) + '\n')
    start = end

fo.close()
```

```python
fo = open(inalnfile, 'w')
for strain in seqDict:
    fo.write('>' + strain + '\n')
    for og in seqDict[strain]:
        fo.write(seqDict[strain][og])
    fo.write('\n')

fo.close()
```
