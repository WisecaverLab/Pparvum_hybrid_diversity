```python
from Bio import SeqIO
import subprocess
```

```python
trimdir="hybrid_orthofinder/19_trimal"
outdir="hybrid_orthofinder/20_long_alns"
fastadir="hybrid_orthofinder/16_orthofinder_fasta"
min_length=150

count_strains_cmd="ls " + fastadir
count=subprocess.check_output(count_strains_cmd, shell=True).decode().split("\n")
count.remove('')
number_strains=len(count)

lscmd="ls " + trimdir + "/*.aln"
files=subprocess.check_output(lscmd, shell=True).decode().split("\n")
files.remove('')
```

```python
to_copy_over=[]
for file in files:
    all_records=SeqIO.parse(file, 'fasta')
    counter=0
    keep=True
    for record in all_records:
        counter += 1
        length=len(str(record.seq))
        if length > min_length and counter == number_strains:
            cpcmd="cp " + file + " " + outdir + "/" + file.split("/")[6]
            subprocess.run(cpcmd, shell=True)
        
```
