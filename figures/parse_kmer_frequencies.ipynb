```python
import pandas as pd
from glob import glob
```

```python
infiles = '../../../hybrid_diversity/3_kmer_frequencies/*.histo'
outfile = 'kmer_freq_allstrains.Rin'
mincov = 10
maxcov = 120
```

```python
typeDict = {
    "8020": "HIGH",
    "8027": "HIGH",
    "12A1": "HIGH",
    "12A1-SC1": "HIGH",
    "12A1-SC2": "HIGH",
    "12A1-SC3": "HIGH",
    "12B1": "LOW",
    "13A5": "HIGH",
    "ARC66": "HIGH",
    "ARC81": "HIGH",
    "ARC82": "HIGH",
    "ARC83": "HIGH",
    "ARC84": "HIGH",
    "CC2192B4": "LOW",
    "CC2192B5": "LOW",
    "CC2192C1": "LOW",
    "CC2192D2": "LOW",
    "CC2192D3": "LOW",
    "CCMP2941": "LOW",
    "CCMP3037": "LOW",
    "NIES1812": "LOW",
    "PK2297D4": "HIGH",
    "PK2298B4": "LOW",
    "PK2298C6": "HIGH",
    "PK2299B3": "HIGH",
    "PK2300D3": "HIGH",
    "RCC3703": "LOW",
    "UTEX2797": "HIGH",
    "UTEX2797-SC1": "HIGH",
    "UTEX2797-SC2": "HIGH",
    "UTEX2797-SC3": "HIGH",
    "UTEX2797-SC4": "HIGH",
    "UTEX2827": "HIGH"

}
```

```python
files = glob(infiles)

dfbase = pd.DataFrame(columns = ['cov','freq','strain'])

for file in files:
    strain = file.split("/")[-1].split("_")[0]
    print(strain)
    if strain not in typeDict:
        print(strain)
        continue
    
    df = pd.read_csv(file, sep = "\t", names = ['cov','freq'])
    df = df[df['cov'] > mincov]
    df = df[df['cov'] < maxcov]
    
    df['strain'] = [strain] * len(df)
    dfbase = pd.concat([dfbase, df], ignore_index = True)

    
dfbase.freq = dfbase.freq.astype(int)
dfbase['cov'] = dfbase['cov'].astype(int)
```

```python
dfbase.to_csv(outfile, sep="\t", index=False)
```
