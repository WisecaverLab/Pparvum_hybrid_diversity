```python
import glob
```

```python
infiles = '../../../hybrid_diversity/4_read_mapping/3_deduplicated/2_nuc_and_chloroplast/*.log2ratio.tsv'
outfile = '../../../hybrid_diversity/4_read_mapping/3_deduplicated/2_nuc_and_chloroplast/log2ratios_combined.tsv'

strainlist = ['UTEX2797', 'PK2298C6', 'PK2300D3', '8020', 'ARC66', 'ARC81', 'ARC82', 'ARC83', 'ARC84', 'UTEX2827', 'PK2297D4', 'PK2299B3', '8027', '13A5', '12A1']
```

```python
fo = open(outfile, 'w')
fo.write('strain\tchr\tstart\tend\tlog2ratio\n')
for infile in glob.glob(infiles):
    strain = infile.split('/')[-1].split('.')[0]
    if strain in strainlist:
        print(strain)
        fi = open(infile)

        for line in fi:
            if line.startswith('chr'):
                continue
            fo.write(strain + '\t' + line)

        fi.close()
fo.close()        
```
