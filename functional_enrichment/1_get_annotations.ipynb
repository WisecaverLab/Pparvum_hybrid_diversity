```python
infile = '../../../genome_diversity/6_Genome_annotation/7_Functional_Annotations/all_annotations_kinfin.txt'
gofile = '../../../hybrid_diversity/7_functional_enrichment/go_ips_annotations.txt'
keggfile = '../../../hybrid_diversity/7_functional_enrichment/kegg_kofs_annotations.txt'
```

```python
with open(infile, 'r') as fi, open(gofile, 'w') as fog, open(keggfile, 'w') as fok:
    for line in fi:
        col = line.rstrip().split('\t')
        if col[0].endswith("_12B1"):
            #print(line)
            go = col[2]
            kegg = col[3]
            newid = '12B1' + col[0].split('.')[0]
            #print(newid, go, kegg)
            
            if go != 'None':
                fog.write(f'{newid}\t{go}\n')
                
            if kegg != 'None':
                fok.write(f'{newid}\t{kegg}\n')
```

```python
infile = '../../../hybrid_diversity/7_functional_enrichment/eggnog/02_eggnog.emapper.annotations'
goenfile = '../../../hybrid_diversity/7_functional_enrichment/go_eggnog_annotations.txt'
keggenfile = '../../../hybrid_diversity/7_functional_enrichment/kegg_eggnog_annotations.txt'
```

```python
with open(infile, 'r') as fi, open(goenfile, 'w') as fog, open(keggenfile, 'w') as fok:
    for line in fi:
        col = line.rstrip().split('\t')
        #print(line)
        if len(col) < 12:
            continue
        
        go = col[9]
        go = go.replace(',', ';')
        kegg = col[11]
        kegg = kegg.replace(',', ';')
        kegg = kegg.replace('ko:', '')
        #print(col[0], go, kegg)

        if go != '-':
            fog.write(f'{col[0]}\t{go}\n')

        if kegg != '-':
            fok.write(f'{col[0]}\t{kegg}\n')
```
