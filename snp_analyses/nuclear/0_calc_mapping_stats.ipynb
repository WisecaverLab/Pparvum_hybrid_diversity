```bash
for bam in *_to_12B1_NCBI_bwa.sorted.bam; do
  strain=$(basename $bam _markduplicates.bam)
  samtools flagstat $bam | awk -v s=$strain '
    /primary$/ {total=$1}
    /primary mapped/ {mapped=$1}
    END {
      print s"\ttotal="total"\tmapped="mapped"\tmap_rate="mapped/total
    }'
done
```
